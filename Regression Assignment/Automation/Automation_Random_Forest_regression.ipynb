```python
import pandas as pd
```

```python
dataset = pd.read_csv("insurance_pre.csv")
```

```python
dataset = pd.get_dummies(dataset, drop_first = 'true')
dataset = dataset.astype(int)
```

```python
dataset.columns
```

```python
independent=dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
dependent = dataset[['charges']]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(independent, dependent, test_size=0.30, random_state = 0)
```

```python
#https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.RandomForestRegressor.html
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score

criteria = ['squared_error', 'absolute_error', 'poisson']
max_features_values = [1.0, 'sqrt', 'log2']
n_estimators_values = [10, 50, 100]

best_r2 = float('-inf')
best_parameters = None

for criterion in criteria:
    for max_features in max_features_values:
        for n_estimators in n_estimators_values:

            regressor = RandomForestRegressor(
                criterion=criterion,
                max_features=max_features,
                n_estimators=n_estimators,
                random_state=0
            )

            # Convert Y_train to 1D
            regressor.fit(X_train, Y_train.values.ravel())

            y_pred = regressor.predict(X_test)

            r2 = r2_score(Y_test, y_pred)

            print(
                "Criterion:", criterion,
                "| Max_features:", max_features,
                "| n_estimators:", n_estimators,
                "| R²:", round(r2, 4)
            )

            if r2 > best_r2:
                best_r2 = r2
                best_parameters = {
                    'criterion': criterion,
                    'max_features': max_features,
                    'n_estimators': n_estimators
                }

print("\n==============================")
print("BEST RESULT")
print("==============================")
print("Best Criterion:", best_parameters['criterion'])
print("Best Max_features:", best_parameters['max_features'])
print("Best n_estimators:", best_parameters['n_estimators'])
print("Best R² Score:", round(best_r2, 4))
```
