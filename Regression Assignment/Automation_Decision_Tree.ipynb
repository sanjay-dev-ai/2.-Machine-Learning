```python
import pandas as pd
```

```python
dataset = pd.read_csv("insurance_pre.csv")
```

```python
dataset = pd.get_dummies(dataset, drop_first = 'true')
dataset = dataset.astype(int)
```

```python
dataset.columns
```

```python
independent=dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
```

```python
dependent = dataset[['charges']]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(independent, dependent, test_size=0.30, random_state = 0)
```

```python
#https://scikit-learn.org/stable/modules/generated/sklearn.tree.DecisionTreeRegressor.html
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score

criteria = ['squared_error', 'absolute_error', 'poisson']
splitters = ['best', 'random']
max_features_values = [None, 'sqrt', 'log2']

best_r2 = float('-inf')
best_parameters = None

for criterion in criteria:
    for splitter in splitters:
        for max_features in max_features_values:

            regressor = DecisionTreeRegressor(
                criterion=criterion,
                splitter=splitter,
                max_features=max_features,
                random_state=0
            )

            regressor.fit(X_train, Y_train)

            y_pred = regressor.predict(X_test)

            r2 = r2_score(Y_test, y_pred)

            print(
                "Criterion:", criterion,
                "| Splitter:", splitter,
                "| Max_features:", max_features,
                "| R²:", round(r2, 4)
            )

            if r2 > best_r2:
                best_r2 = r2
                best_parameters = {
                    'criterion': criterion,
                    'splitter': splitter,
                    'max_features': max_features
                }

print("\n==============================")
print("BEST RESULT")
print("==============================")
print("Best Criterion:", best_parameters['criterion'])
print("Best Splitter:", best_parameters['splitter'])
print("Best Max_features:", best_parameters['max_features'])
print("Best R² Score:", round(best_r2, 4))
```
