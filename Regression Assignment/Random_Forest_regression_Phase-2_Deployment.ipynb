```python
import pickle

load_model = pickle.load(open('final_Model_Insurance_Charges_Predict.sav', 'rb'))

age = int(input("Enter your Age: "))
bmi = float(input("Enter your BMI: "))
children = int(input("Enter No. of Children: "))

sex = input("Enter your Sex (male/female): ").lower()
smoker = input("Are you a Smoker? (yes/no): ").lower()

# Convert categorical values into numerical values
if sex == "male":
    sex_male = 1
elif sex == "female":
    sex_male = 0
else:
    print("Invalid input for sex. Please enter male or female.")

if smoker == "yes":
    smoker_yes = 1
elif smoker == "no":
    smoker_yes = 0
else:
    print("Invalid input for smoker. Please enter yes or no.")

# Prediction
result = load_model.predict(
    [[age, bmi, children, sex_male, smoker_yes]]
)

print("Predicted Insurance Charges:", result[0])
```
