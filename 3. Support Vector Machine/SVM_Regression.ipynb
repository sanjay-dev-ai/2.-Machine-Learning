```python
import pandas as pd
```

```python
dataset = pd.read_csv("50_Startups.csv")
```

```python
dataset
```

```python
dataset = pd.get_dummies(dataset, drop_first=True)
dataset = dataset.astype(int)
dataset
```

```python
dataset.columns
```

```python
independent = dataset[['R&D Spend', 'Administration', 'Marketing Spend', 'State_Florida', 'State_New York']]
```

```python
dependent = dataset[['Profit']]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(independent, dependent, test_size = 0.30, random_state = 0)
```

```python
from sklearn.preprocessing import StandardScaler   #standardizing the data using the values of (mean, standard deviation formula)
sc=StandardScaler()
X_train = sc.fit_transform(X_train)             #fit method does only calcaulation, Where as transform will do the calculation and provides the final value 
X_test = sc.transform(X_test)
```

```python
X_train
```

```python
from sklearn.svm import SVR
regressor = SVR(kernel = "rbf", C=10)     #The "C" is known as parameter where number of error allowed to create a model
regressor.fit(X_train, Y_train)

#Kernal refers to the algorithm used to create a model in SVM (Support Vector Machine). 
#Types of algorithm - rbf (radial bias function), kernel{‘linear - Straight Line’, ‘poly - Parabola’, ‘rbf - Circle’, ‘sigmoid - S Shaped ’, ‘precomputed’} or callable, default=’rbf’
```

```python
regressor.intercept_
```

```python
regressor.n_support_          #Verifying the number of support vectors (datapoints) used to create model
```

```python
regressor.support_            #List of support vectors
```

```python
y_pred = regressor.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r2_score = r2_score(Y_test, y_pred)
```

```python
r2_score
```
